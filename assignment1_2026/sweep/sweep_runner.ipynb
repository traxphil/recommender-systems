| file | content |
|---|---|
| `sweep_queue.json` | configs to run (new rounds are appended while the sweep runs) |

| `sweep_epoch_history.csv` | train / valid MSE of every epoch of every run |
| `sweep_log.txt` | everything the training prints |

Run all cells once. Cell 3 keeps running and picks up new configs from the queue;
